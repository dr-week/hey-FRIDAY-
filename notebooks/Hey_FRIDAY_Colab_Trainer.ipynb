# 🧠 Hey FRIDAY: Free Google Colab GPU Trainer
Train custom zero-latency intent embeddings from your Google Drive `HeyFRIDAY-Training` feedback logs.

```python
# 1. Mount Google Drive
from google.colab import drive
import os, json, glob
drive.mount('/content/drive')
training_dir = '/content/drive/MyDrive/HeyFRIDAY-Training'
print(f"Searching for training files in {training_dir}...")
files = glob.glob(f"{training_dir}/*.jsonl")
print(f"Found {len(files)} dataset files.")
```

```python
# 3. Load user-verified speech pairs
data = []
for f in files:
    with open(f) as fp:
        for line in fp:
            item = json.loads(line)
            if 'rawSpokenText' in item or 'reframed' in item:
                query = item.get('correctedText') or item.get('reframed') or item.get('rawSpokenText')
                intent = item.get('detectedIntent') or item.get('intent')
                if query and intent:
                    data.append((query, intent))
print(f"Loaded {len(data)} training samples from Hey FRIDAY feedback.")
```
